Trying out the MatBench is-a-metal Task.

```python
from matminer.datasets import load_dataset
from sklearn.model_selection import StratifiedKFold
from CBFV import composition
import pandas as pd
```

Loading in dataset

```python
df = load_dataset("matbench_expt_is_metal")
df.columns = ["formula", "class"]
```

CBFV generation

```python
CBFVs = ["magpie", "oliynyk", "mat2vec"]
combined = pd.DataFrame()
for cbfv in CBFVs:
    df_copy = df.copy()
    df_copy = df_copy.drop("class", axis=1)
    df_copy["target"] = 0
    X, y, formulae, skipped = composition.generate_features(df_copy,
                                                            elem_prop=cbfv,
                                                            drop_duplicates=True,
                                                            extend_features=True,
                                                            sum_feat=True)
    combined = pd.concat([combined, X], axis=1)
    X["formula"] = df["formula"]
    X["class"] = df["class"]
    X.to_csv(f"data_{cbfv}.csv", index=False)
combined["formula"] = df_copy["formula"]
combined["class"] = df["class"]
combined.to_csv(r"data_allcbfv.csv")
```
