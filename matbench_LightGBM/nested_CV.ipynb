```python
import pickle as pkl
import numpy as np
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
summary_path = r"C:\Users\eli827\OneDrive - The University of Liverpool\Ceres Power Data Cleaning and Modelling\matbench_LightGBM\matbench_results\classifier_summary.pkl"
with open(summary_path, "rb") as f: 
    summary = pkl.load(f)

all_y_true, all_y_pred = summary["fold_results"]["all_y_true"], summary["fold_results"]["all_y_pred"]
cm = confusion_matrix(all_y_true, all_y_pred, labels=[1, 0])
cm = np.asarray(cm, dtype=np.int64)

fig, ax = plt.subplots(figsize=(4, 4))
ConfusionMatrixDisplay(cm, display_labels=["metal", "non-metal"]).plot(
    cmap="Blues", values_format="d", ax=ax, colorbar=False,
)
plt.show()
```
