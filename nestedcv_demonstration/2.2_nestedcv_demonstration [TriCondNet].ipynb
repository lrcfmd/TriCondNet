# TriCondNet,  nested cross-validation example

Minimal end-to-end run of the TriCondNet nested cross-validation pipeline. A
LightGBM metal/semiconductor classifier routes each composition to a
physics-informed regressor (`MetalPINN` or `SemiconductorPINN`), with each
network's hyperparameters tuned by Bayesian optimisation within inner folds.


```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent))
from evaluation.NestedKFold_TriCondNet import TriCondNet_NestedKFold_Pipeline
import warnings
# Supressing cosmetic warnings from LightGBM
warnings.filterwarnings("ignore", message="X does not have valid feature names")
```

```python
results_dir = "nested_cv_results"
data_path = Path.cwd().parent / "data" / "magpie_data.csv"

n_inner        = 5
n_outer        = 5
num_opt_trials = 25
random_state   = 100

pipeline = TriCondNet_NestedKFold_Pipeline(
    data_path=data_path,
    cbfv="magpie",
    n_inner=n_inner,
    n_outer=n_outer,
    num_opt_trials=num_opt_trials,
    target_name="target",
    random_state=random_state,
    file_type="csv",
)

fold_results, all_hyperparameters = pipeline.run_nested_cv(results_dir=results_dir, retrain=True)
```

Parity plot of predicted vs. actual values. Points should fall as close as possible to the y = x diagonal, indicating agreement between prediction and ground truth.
