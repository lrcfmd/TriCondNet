# TriMODNet nested cross-validation example

Same nested cross-validation setup as TriCondNet, but the LightGBM classifier and the two physics-informed regressors are all replaced by MODNet models (neural-network based).

Update `data_path` to point at your CBFV feature CSV before running.

```python
import os, warnings, logging
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
import tensorflow as tf
tf.get_logger().setLevel("ERROR")
tf.autograph.set_verbosity(0)
from tensorflow.python.util import deprecation
deprecation._PRINT_DEPRECATION_WARNINGS = False 

import absl.logging
absl.logging.set_verbosity(absl.logging.ERROR)
warnings.filterwarnings("ignore", category=RuntimeWarning)
logging.getLogger("tensorflow").setLevel(logging.ERROR)
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent))
from evaluation.NestedKFold_TriMODNet import TriMODNet_NestedKFold_Pipeline
```

```python
results_dir = "nested_cv_results"
data_path = Path.cwd().parent / "data" / "magpie_data.csv"

n_inner        = 5
n_outer        = 5
num_opt_trials = 25
random_state   = 100

Pipeline_TMN = TriMODNet_NestedKFold_Pipeline(
    data_path=data_path,
    cbfv="magpie",
    n_inner=n_inner,
    n_outer=n_outer,
    num_opt_trials=num_opt_trials,
    target_name="target",
    random_state=random_state,
)

fold_results, all_hyperparameters = Pipeline_TMN.run_nested_cv(retrain=True, results_dir=results_dir)
```
